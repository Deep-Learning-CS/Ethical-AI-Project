```python
import numpy as np

# Define BIO labels and mapping (if not already defined in a previous cell)
label_list = ['O', 'B-MANIP', 'I-MANIP']
label2id = {label: i for i, label in enumerate(label_list)}
id2label = {i: label for i, label in enumerate(label_list)}

# Function to convert character offsets to token BIO labels (CORRECTED VERSION)
def convert_offsets_to_bio(example):
    # Tokenize the content and get offset mapping
    tokenized_inputs = tokenizer(
        example["content"],
        truncation=True,
        max_length=tokenizer.model_max_length,
        return_offsets_mapping=True,
        padding=False
    )

    offset_mapping = tokenized_inputs["offset_mapping"]
    trigger_spans = example["trigger_words"] # Get the trigger words list

    # === CORRECTION START ===
    # Check if trigger_spans is None and default to an empty list if it is
    if trigger_spans is None:
        trigger_spans = []
    # === CORRECTION END ===

    # Initialize labels
    labels = np.zeros(len(offset_mapping), dtype=int)

    # Sort trigger spans if the list is not empty
    if trigger_spans: # Only sort if there are spans
        trigger_spans.sort(key=lambda x: x[0])

    span_idx = 0
    for token_idx, (start_char_token, end_char_token) in enumerate(offset_mapping):
        # Ignore special tokens and padding
        if start_char_token == end_char_token:
            labels[token_idx] = -100
            continue

        # Check overlap with trigger spans (only if there are spans to check)
        if trigger_spans:
             while span_idx < len(trigger_spans) and start_char_token >= trigger_spans[span_idx][1]:
                  span_idx += 1

             if span_idx >= len(trigger_spans):
                  # No more spans to check for this example
                  # The break was removed to ensure the -100 logic below runs for all tokens
                  pass # Let the loop continue to handle subword labeling
             else:
                  current_span_start, current_span_end = trigger_spans[span_idx]
                  # Check for overlap
                  if start_char_token < current_span_end and end_char_token > current_span_start:
                       # Determine if it's B-MANIP or I-MANIP
                       is_beginning = True
                       if token_idx > 0:
                            prev_start_char, prev_end_char = offset_mapping[token_idx - 1]
                            # Check if previous token was special OR also within the current span
                            if prev_start_char == prev_end_char or (prev_start_char >= current_span_start and prev_end_char <= current_span_end):
                                 is_beginning = False

                       # Refined check: Is the token's start >= span's start AND is it the first token we've seen for this span?
                       is_truly_beginning = (start_char_token >= current_span_start) and is_beginning

                       if is_truly_beginning:
                            labels[token_idx] = label2id['B-MANIP']
                       else:
                            labels[token_idx] = label2id['I-MANIP']

        # else: if not trigger_spans, all tokens remain 'O' (which is label 0) initialized above

    # Assign -100 to subword continuations and special tokens
    word_ids = tokenized_inputs.word_ids()
    previous_word_idx = None
    final_labels = []
    for i, token_label in enumerate(labels):
        # Use offset_mapping to identify special tokens reliably
        start_char, end_char = offset_mapping[i]
        if start_char == end_char: # Handle special tokens like [CLS], [SEP]
            final_labels.append(-100)
            continue

        word_idx = word_ids[i]
        if word_idx is None: # Should be covered by offset check, but good safety net
            final_labels.append(-100)
        elif word_idx != previous_word_idx: # First token of a new word
            final_labels.append(token_label)
        else: # Subword continuation token
            # Assign -100 to be ignored in loss, *unless* the task requires labeling subwords (unlikely for BIO)
            # Check if the BIO label is I-MANIP, potentially keep it? Depends on task specifics.
            # Standard practice: ignore subword continuations.
            final_labels.append(-100)
        previous_word_idx = word_idx

    tokenized_inputs["labels"] = final_labels
    return tokenized_inputs
```

```python
# Make sure raw_datasets is loaded from the previous step
if 'raw_datasets' in locals():
    # Apply the function - this might take a few minutes
    tokenized_datasets = raw_datasets.map(
        convert_offsets_to_bio,
        batched=False, # Process example by example for offset logic clarity
        remove_columns=raw_datasets['train'].column_names # Remove old columns to avoid conflicts
    )

    print("\nDataset after tokenization and label conversion:")
    print(tokenized_datasets)

    # Check the first example again to see the new structure
    print("\nFirst example after processing:")
    if len(tokenized_datasets['train']) > 0:
        example = tokenized_datasets['train'][0]
        print(f"Tokens (IDs): {example['input_ids'][:50]}...") # Show some token IDs
        print(f"Labels (IDs): {example['labels'][:50]}...") # Show corresponding label IDs
        # Optional: Decode to see tokens and labels
        tokens = tokenizer.convert_ids_to_tokens(example['input_ids'])
        label_names = [id2label.get(l, 'IGN') for l in example['labels']] # IGN for -100
        print("\nDecoded Tokens and Labels (first 50):")
        for token, label in zip(tokens[:50], label_names[:50]):
              print(f"{token:<15} {label}")

    else:
        print("Processed training set is empty.")

else:
    print("Raw dataset not loaded. Please run the loading cell first.")
```

```python
# === Cleaned Up Step 6.2: Define RoBERTa + CRF Custom Model Class ===
import torch
import torch.nn as nn
from transformers import AutoModel, PreTrainedModel, RobertaConfig
# Use torchcrf if that worked for you
from torchcrf import CRF
from transformers.modeling_outputs import TokenClassifierOutput

# Ensure label mappings are defined (e.g., from Step 4)
# label_list = ['O', 'B-MANIP', 'I-MANIP']
# label2id = {'O': 0, 'B-MANIP': 1, 'I-MANIP': 2}
# id2label = {0: 'O', 1: 'B-MANIP', 2: 'I-MANIP'}
num_labels = len(label_list)

class RobertaCrfForTokenClassification(PreTrainedModel):
    config_class = RobertaConfig

    # --- Uses checkpoint passed during instantiation ---
    def __init__(self, config, checkpoint):
        super().__init__(config)
        self.num_labels = config.num_labels

        # --- Load RoBERTa base model using the passed checkpoint name ---
        self.roberta = AutoModel.from_pretrained(
            checkpoint,
            config=config,
            add_pooling_layer=False
        )
        # --- End of change ---

        self.dropout = nn.Dropout(config.hidden_dropout_prob if hasattr(config, 'hidden_dropout_prob') else 0.1)
        self.classifier = nn.Linear(config.hidden_size, self.num_labels)
        # Ensure batch_first=True if your library version supports/defaults to it
        self.crf = CRF(self.num_labels, batch_first=True)


    def forward(
        self,
        input_ids=None,
        attention_mask=None,
        position_ids=None,
        head_mask=None,
        inputs_embeds=None,
        labels=None,
        output_attentions=None,
        output_hidden_states=None,
        return_dict=None,
    ):
        return_dict = return_dict if return_dict is not None else self.config.use_return_dict
        outputs = self.roberta(
            input_ids,
            attention_mask=attention_mask,
            position_ids=position_ids,
            head_mask=head_mask,
            inputs_embeds=inputs_embeds,
            output_attentions=output_attentions,
            output_hidden_states=output_hidden_states,
            return_dict=True,
        )
        sequence_output = outputs.last_hidden_state
        sequence_output = self.dropout(sequence_output)
        logits = self.classifier(sequence_output) # Shape: (batch_size, seq_length, num_labels)

        loss = None
        if labels is not None:
             # Use byte mask based on previous findings, check library docs if error persists
             crf_mask = attention_mask.byte() if attention_mask is not None else None

             # Mask labels for CRF input (replace -100 with 'O', ensure padding is also 'O')
             masked_labels = labels.clone()
             if crf_mask is not None:
                 # Use boolean mask for indexing based on crf_mask
                 bool_mask_for_indexing = crf_mask.bool()
                 masked_labels[~bool_mask_for_indexing] = label2id['O'] # Set padding tokens to 'O'
             masked_labels[labels == -100] = label2id['O'] # Set originally ignored subwords to 'O'

             # --- Debugging Prints Removed ---

             # Calculate CRF loss
             # Ensure masked_labels is LongTensor for CRF tags input
             loss = -self.crf(emissions=logits, tags=masked_labels.long(), mask=crf_mask, reduction='mean')

        return TokenClassifierOutput(
            loss=loss,
            logits=logits, # Return logits for compute_metrics compatibility
            hidden_states=outputs.hidden_states,
            attentions=outputs.attentions,
        )

print("Custom RobertaCrfForTokenClassification class defined (Clean Version).")
```

```python
# === Corrected New Step 6.3: Define RoBERTa Config and Instantiate Custom Model ===
from transformers import RobertaConfig

# Ensure label mappings and num_labels are defined
# Ensure tokenizer_checkpoint = "youscan/ukr-roberta-base" is defined in Step 3

# Define config using the RoBERTa checkpoint name
config = RobertaConfig.from_pretrained(
    tokenizer_checkpoint,
    num_labels=num_labels,
    id2label=id2label,
    label2id=label2id
)

# Instantiate the RoBERTa custom model, passing the checkpoint name
model = RobertaCrfForTokenClassification(
    config=config,
    checkpoint=tokenizer_checkpoint # <<< PASS the checkpoint string
)
print("Custom RoBERTa+CRF model instantiated.")
```

```python
# === Rerun Step 7 ===
from transformers import DataCollatorForTokenClassification # Ensure import if needed
data_collator = DataCollatorForTokenClassification(tokenizer=tokenizer)
print("Data collator ready (using mBERT tokenizer).")
```

```python
import numpy as np
import evaluate # Make sure evaluate and seqeval are installed

# Load the seqeval metric
seqeval_metric = evaluate.load("seqeval")

# Define the function to compute metrics during evaluation
def compute_metrics(p):
    predictions, labels = p
    # Get the most likely label ID for each token prediction
    predictions = np.argmax(predictions, axis=2)

    # Remove ignored index (-100) and convert IDs to label strings
    true_predictions = [
        [id2label[p] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]
    true_labels = [
        [id2label[l] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]

    # Compute metrics using seqeval
    results = seqeval_metric.compute(predictions=true_predictions, references=true_labels)

    # Return metrics including overall F1 (the primary metric for seqeval)
    return {
        "precision": results["overall_precision"],
        "recall": results["overall_recall"],
        "f1": results["overall_f1"],
        "accuracy": results["overall_accuracy"],
    }

print("Compute_metrics function defined.")
```

```python
# === Modified Step 9 ===
from transformers import TrainingArguments # Ensure import if needed

# Define output directory for CRF results
output_dir = "./unlp-span-detection-results-ukr-roberta-crf" # <<< CHANGED NAME

# Define training arguments
training_args = TrainingArguments(
    output_dir=output_dir,
    evaluation_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3, # <<< SET TO 3 (or keep 8 if preferred) >>>
    weight_decay=0.01,
    logging_dir=f"{output_dir}/logs",
    logging_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,
    report_to="none",
)
print("Training arguments updated for CRF run.")
```

```python
# === Modified Step 10 ===
from transformers import Trainer # Ensure import if needed

# Ensure train/eval split uses the mBERT-processed tokenized_datasets
if 'validation' not in tokenized_datasets:
    print("Validation split not found. Splitting train set (adjust ratio if needed)...")
    split_dataset = tokenized_datasets['train'].train_test_split(train_size=int(0.85 * len(tokenized_datasets['train'])), seed=42)
    train_dataset = split_dataset['train']
    eval_dataset = split_dataset['test']
    print(f"Train examples: {len(train_dataset)}, Validation examples: {len(eval_dataset)}")
else:
    train_dataset = tokenized_datasets['train']
    eval_dataset = tokenized_datasets['validation']
    print(f"Using existing train ({len(train_dataset)}) and validation ({len(eval_dataset)}) sets.")

# Initialize the Trainer with the CUSTOM model
trainer = Trainer(
    model=model,                          # <<< USES CUSTOM CRF MODEL >>>
    args=training_args,                   # Updated args
    train_dataset=train_dataset,          # mBERT processed data
    eval_dataset=eval_dataset,            # mBERT processed data
    tokenizer=tokenizer,                  # mBERT tokenizer
    data_collator=data_collator,          # mBERT collator
    compute_metrics=compute_metrics,      # Same metrics function
)
print("Trainer initialized with custom CRF model.")
```

```python
# Start the training process
print("Starting training...")
train_result = trainer.train()
print("Training finished!")

# Optional: Save metrics from the training process
metrics = train_result.metrics
trainer.log_metrics("train", metrics)
trainer.save_metrics("train", metrics)

# Save the final best model (although load_best_model_at_end=True should have loaded it)
# This saves the tokenizer and config alongside the model weights
print("Saving the best model...")
trainer.save_model(f"{output_dir}/best_model") # Save to a subdirectory
print(f"Best model saved to {output_dir}/best_model")
```
